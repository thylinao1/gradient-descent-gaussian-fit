# Fitting a Gaussian Model to Height Data (Built From Scratch)

In this project, I implemented a **Gaussian fitting algorithm from scratch** to model the distribution of human heights using gradient descent (steepest descent).  
I didn’t use any ML libraries for optimization, everything here is built manually using **NumPy** and **Matplotlib**.



## Step 1: Load Tools and Data

I start by importing the required libraries and a small helper module that provides the height dataset and visualization utilities.

```
from readonly.HeightsModule import *
import numpy as np
import matplotlib.pyplot as plt
```

## Step 2: Define the Gaussian Function

The height distribution can be modeled as a **Gaussian (normal) distribution)**:

\[
f(x;\mu,\sigma)=\frac{1}{\sqrt{2\pi}\,\sigma}\exp\!\left(-\frac{(x-\mu)^2}{2\sigma^2}\right)
\]

This function defines the bell-shaped curve based on parameters μ (mean) and σ (spread).

```
def f(x, mu, sig):
    """Gaussian probability density function."""
    return np.exp(-(x - mu)**2 / (2 * sig**2)) / (np.sqrt(2 * np.pi) * sig)
```

## Step 3: Compute Analytical Derivatives

To optimize the parameters with gradient descent, I derived the analytical derivatives of the Gaussian with respect to μ and σ:

\[
\frac{\partial f}{\partial \mu} = f(x;\mu,\sigma)\frac{x-\mu}{\sigma^2}
\]

\[
\frac{\partial f}{\partial \sigma} = f(x;\mu,\sigma)\left(-\frac{1}{\sigma} + \frac{(x-\mu)^2}{\sigma^3}\right)
\]

```
def dfdmu(x, mu, sig):
    """Partial derivative of f with respect to mu."""
    return f(x, mu, sig) * (x - mu) / sig**2

def dfdsig(x, mu, sig):
    """Partial derivative of f with respect to sigma."""
    return f(x, mu, sig) * (-(1 / sig) + ((x - mu)**2 / sig**3))
```

## Step 4: Implement Gradient Descent

Here I define one optimization step that moves μ and σ opposite to the gradient of the chi-square error:

\[
\chi^2 = \sum (y - f(x; \mu, \sigma))^2
\]

This custom function performs a single steepest-descent step.

```
def steepest_step(x, y, mu, sig, learning_rate):
    """Perform one step of gradient descent to update mu and sigma."""
    J_mu = -2 * (y - f(x, mu, sig)) @ dfdmu(x, mu, sig)
    J_sig = -2 * (y - f(x, mu, sig)) @ dfdsig(x, mu, sig)
    mu -= learning_rate * J_mu
    sig -= learning_rate * J_sig
    return mu, sig
```

## Step 5: Run the Optimization Loop

Now I initialize μ and σ, choose a learning rate, and run several iterations to see how the parameters evolve.

```
x, y = heights_data()
mu, sig = 155, 6
path = [(mu, sig)]

for i in range(50):
    mu, sig = steepest_step(x, y, mu, sig, learning_rate=2000)
    path.append((mu, sig))
```

## Step 6: Visualize the Results

To analyze the fit, I visualize both the histogram and the optimization path.

```
histogram(f, [mu, sig])
contour(f, np.array(path))
histogram(f, [mu, sig])
```
